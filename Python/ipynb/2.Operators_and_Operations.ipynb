Python Operators

In Python, Operators are the symbols used to perform a specific operation on different values and variables. These values and variables are considered as the Operands, on which the operator is applied.


Arithmetic Operators (Perform basic math operations)
| Operator | Operation      | Example  | Result |
| -------- | -------------- | -------- | ------ |
| `+`      | Addition       | `3 + 2`  | `5`    |
| `-`      | Subtraction    | `5 - 2`  | `3`    |
| `*`      | Multiplication | `3 * 4`  | `12`   |
| `/`      | Division       | `10 / 2` | `5.0`  |
| `//`     | Floor Division | `7 // 2` | `3`    |
| `%`      | Modulus        | `7 % 2`  | `1`    |
| `**`     | Exponentiation | `2 ** 3` | `8`    |

Comparison (Relational) Operators (Compare values)
| Operator | Operation           | Example  | Result  |
| -------- | ------------------- | -------- | ------- |
| `==`     | Equal to            | `5 == 5` | `True`  |
| `!=`     | Not equal to        | `4 != 3` | `True`  |
| `>`      | Greater than        | `7 > 2`  | `True`  |
| `<`      | Less than           | `3 < 1`  | `False` |
| `>=`     | Greater or equal to | `5 >= 5` | `True`  |
| `<=`     | Less or equal to    | `2 <= 3` | `True`  |

Logical Operators (Used for boolean logic)
| Operator | Operation   | Example          | Result  |
| -------- | ----------- | ---------------- | ------- |
| `and`    | Logical AND | `True and False` | `False` |
| `or`     | Logical OR  | `True or False`  | `True`  |
| `not`    | Logical NOT | `not True`       | `False` |

Assignment Operators (Assign values to variables)
| Operator | Operation               | Example   | Equivalent To |
| -------- | ----------------------- | --------- | ------------- |
| `=`      | Assign                  | `x = 5`   |               |
| `+=`     | Add and assign          | `x += 3`  | `x = x + 3`   |
| `-=`     | Subtract and assign     | `x -= 2`  | `x = x - 2`   |
| `*=`     | Multiply and assign     | `x *= 4`  | `x = x * 4`   |
| `/=`     | Divide and assign       | `x /= 2`  | `x = x / 2`   |
| `//=`    | Floor divide and assign | `x //= 3` | `x = x // 3`  |
| `%=`     | Modulus and assign      | `x %= 2`  | `x = x % 2`   |
| `**=`    | Power and assign        | `x **= 2` | `x = x ** 2`  |

Bitwise Operators (Operate on binary values)
| Operator | Name        | Example  | Description         | 
| -------- | ----------- | -------- | ------------------- | 
| `&`      | AND         | `5 & 3`  | Binary AND          | 
| \`       | \`          | OR       | \`5                 | 
| `^`      | XOR         | `5 ^ 3`  | Binary exclusive OR | 
| `~`      | NOT         | `~5`     | Bitwise NOT         | 
| `<<`     | Left shift  | `5 << 1` | Shift bits left     | 
| `>>`     | Right shift | `5 >> 1` | Shift bits right    | 

Membership Operators (Test membership in a sequence)
| Operator | Operation             | Example            | Result |
| -------- | --------------------- | ------------------ | ------ |
| `in`     | Value in sequence     | `'a' in 'apple'`   | `True` |
| `not in` | Value not in sequence | `'x' not in 'dog'` | `True` |

Identity Operators (Compare memory locations)
| Operator | Operation        | Example      | Result       |
| -------- | ---------------- | ------------ | ------------ |
| `is`     | Same object?     | `x is y`     | `True/False` |
| `is not` | Not same object? | `x is not y` | `True/False` |

Type Casting in Python
