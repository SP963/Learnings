Data Structures: Data Structures are a way of organizing data so that it can be accessed more efficiently depending upon the situation. 

Data Structures in Python: List, Tuple, Dictionary, Set.

Indexing: Indexing is a way to access individual elements in data structures using their position (index). starts form 0 to (n-1)
    positive index -> 0,1,2,3,4,5,....,n-1
    negative index -> -n,-n+1,-n+2,.....-3,-2,-1

Range: used to generate a sequence of numbers. -> range(start, stop, step)

Tuple: Ordered, immutable (cannot be changed after creation), allows duplicates. -> a=(1,2,3)

Dictionary: Unordered ([ maintain insertion order]ordered in Python 3.7+), mutable, key-value pairs. -> a={"A":1,"B":2,"c":3}

Set: Unordered, mutable, no duplicate elements. -> a={1,2,3,4,5}

List: Ordered, mutable (changeable), allows duplicate elements. -> a=[1,2,3]

nested list: List inside another  -> a=[1,2,3,[4,5,6],7,8]

empty list: list with no values -> a=[]

you can access an element from the list using index (either positive or negative) we can perform operations while indexing

Slicing a List: Slicing is a way to extract a subset of elements from a list using a specific pattern. -> list[start:stop:step]

List is mutable so we can add elements to the list -> for adding we use append(), insert(), extend(), + (concatinating 2 lists), * (repeat the list)

Updating Elements into a List: Elements in a list can be updated using indexing.

List is mutable so we can delete elements to the list -> for removing we use remove(value), pop(index), del statement

```python
#creating a list
a=[1,2,3,"c","b","Satya","Mani"]
#Printing a list
print(a)
#Printing a type of list
print(type(a))
#nested list
b=["num","ele",["1","2","3"],"loop"]
c=["z","y","x",a,b,"sum","prod"]
#Printing nested list
print(b)
print(c)
#Printing a type of nested list
print(type(b))
print(type(c))
#accessing element in list using index
print(a[2])
print(a[-2])
#accessing elements in list using index via operations
shopping_list=["Milk","Bread","Chocolate spread","Eggs"]
print(shopping_list[0])
print(shopping_list[4-1])
print(shopping_list[2-1])
print(shopping_list[4+2-5])
#slicing a list
print(a[1:]) #will slice the elements form index 1 to the end (we haven't mentioned the stop value)
print(a[1:3]) #will slice the elements from index 1 to 3
print(a[1:4:2]) #will slice the elements from index 1 to 4 with step size = 2
#adding elements to a list -> using append() it will add at the end
a.append("Satya") #one or more elements 
print(a)
#adding elements to a list -> using insert() for insert you need to give index value when the element need to be added
a.insert(0,"Paddhu")
print(a)
#adding elements to a list -> using extend() it will add at the end need to give list of elements
a.extend(["Anima"])
print(a)
#concatinating 2 lists
m=[1,3,2,4,5,6]
n=[8,7,6,5,4,3]
print(m+n) #adding 2 lists
print(m*2)
# Updating the second element  
a[1] = 52
print(a)  
#removing a value from list
print(c)
c.remove("prod")
print(c)
c.pop(-2)
print(c)
del c[1]
print(c)
```
